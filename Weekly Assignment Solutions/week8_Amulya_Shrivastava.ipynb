## Functional Requirements

The assistant should support:

- `Calculator Tool` – Detect mathematical expressions, evaluate them, and return the computed result.
- `Keyword Extraction Tool` – Extract up to five important keywords from the given text.
- `General Query Handling` – Answer general user queries with an appropriate response.
- `Structured JSON Output` – Return every response in the following format:

```json
{
  "type": "...",
  "result": "..."
}
```

**Tools Used:** `Python`, `Conditional Routing`, `Functions`, `JSON Dictionary Output`, `Basic Exception Handling`

---

##### 🛠️ Code Section 1 — Calculator Tool

````python
def calculator(expression: str) -> str:
    """
    Evaluate a mathematical expression.
    """
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
print("Output 1:",calculator("2 + 3"))
print("Output 2:",calculator("10 * 5"))
print("Output 3:",calculator("(100 - 20) / 4"))
````

##### 🛠️ Code Section 2 — Keyword Extraction Tool

````python
def extract_keywords(text: str) -> list:
    """
    Extract important keywords from text.
    """
    try:
        words = text.split()
        keywords = []
        for word in words:
            word = word.lower().strip(".,!?")
            if len(word) > 4 and word not in keywords:
                keywords.append(word)
        return keywords[:5]
    except Exception:
        return []
print("Output :",extract_keywords("Artificial Intelligence is transforming healthcare through predictive analytics and automation."))
````

##### 🤖 Code Section 3 — Agent Routing Logic

👉 Used conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
import re

def agent(query: str):
    query_lower = query.lower()
    try:
        # Calculator Routing
        if "calculate" in query_lower:
            expression = query_lower.replace("calculate", "").strip()
            result = calculator(expression)
            return {
                "type": "calculation",
                "result": result
            }
        # Keyword Routing
        elif "keyword" in query_lower:
            text = re.sub(
                r"extract\s+keywords\s+from",
                "",
                query,
                flags=re.IGNORECASE
            ).strip()
            result = extract_keywords(text)
            return {
                "type": "keywords",
                "result": result
            }
        # General Queries
        else:
            return {
                "type": "general",
                "result": f"You asked: {query}"
            }
    except Exception as e:
        return {
            "type": "error",
            "result": str(e)
        }
````

##### 🧪 Code Section 5 — Test Cases

````python
queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?",
    "Calculate 50 * 12",
    "Extract keywords from Data Science and Machine Learning are exciting fields"
]
for query in queries:
    print("Query :", query)
    response = agent(query)
    print("Response :", response)
    print("-" * 60, "\n")
````

##### 🎯 Code Section 6 — Interactive Mode

````python
print("Single Agent Smart Assistant")
print("Type 'exit' to stop.\n")

while True:
    user_input = input("Enter query: ")
    if user_input.lower() == "exit":
        print("Assistant Closed.")
        break
    response = agent(user_input)
    print(response)
    print()
````
